# Genotype GQ histogram

Simple histogram of all per-sample genotype quality (GQ) values from
`03_variants/genotype_gq/gq_table.tsv` (output of rule `extract_genotype_gq`).
Use it to pick the `genotype_gq_min` floor for `apply_genotype_gq_filter`.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Path to the GQ table (edit results_dir for your run, e.g. .../results/RL4M4F)
gq_table = "../results/RL4M4F/03_variants/genotype_gq/gq_table.tsv"

# Wide TSV: one column per sample, one row per site. Missing GQ is '.'
df = pd.read_csv(gq_table, sep="\t", na_values=".")
gq = df.to_numpy().ravel()
gq = gq[~pd.isna(gq)]
print(f"{len(gq):,} genotype GQ values across {df.shape[1]} samples")
```

```python
plt.figure(figsize=(8, 4))
plt.hist(gq, bins=50, color="steelblue", edgecolor="white")
plt.axvline(30, color="red", linestyle="--", label="GQ = 30")
plt.xlabel("GQ")
plt.ylabel("genotype count")
plt.title("Genotype quality (GQ) distribution")
plt.legend()
plt.tight_layout()
plt.show()
```
